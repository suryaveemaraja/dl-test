```python

import re
import os
import time
import zipfile
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

print("All imports completed successfully!")
```

```python

clean_texts = []

for text in texts:
    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    if len(text.split()) >= 3:
        clean_texts.append(text)

print("Records after preprocessing:", len(clean_texts))
print("Sample:", clean_texts[:5])

if len(clean_texts) < 10:
    raise ValueError(
        "Dataset is too small. Upload more text data."
    )
```

```python

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences

tokenizer = Tokenizer(
    num_words=10000,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(clean_texts)

vocab_size = min(
    10000,
    len(tokenizer.word_index) + 1
)

sequences = []

for text in clean_texts:
    token_list = tokenizer.texts_to_sequences([text])[0]

    for i in range(2, len(token_list)):
        sequences.append(token_list[:i + 1])

max_sequence_len = min(
    30,
    max(len(seq) for seq in sequences)
)

sequences = [
    seq[-max_sequence_len:]
    for seq in sequences
    if len(seq) >= 3
]

sequences = pad_sequences(
    sequences,
    maxlen=max_sequence_len,
    padding="pre",
    truncating="pre"
)

X = sequences[:, :-1]
y = sequences[:, -1]

print("Vocabulary size:", vocab_size)
print("Input shape:", X.shape)
print("Target shape:", y.shape)
print("Maximum input token:", X.max())
print("Maximum target token:", y.max())

assert X.max() < vocab_size, "Input token exceeds vocabulary"
assert y.max() < vocab_size, "Target token exceeds vocabulary"
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        mask_zero=True
    ),
    LSTM(64),
    Dropout(0.2),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.build(input_shape=(None, X_train.shape[1]))
model.summary()
```

```python

from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

start_time = time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=64,
    validation_split=0.1,
    callbacks=[early_stop],
    verbose=1
)

training_time = time.time() - start_time

print("Training time:", round(training_time, 2), "seconds")
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

error_rate = 1 - test_accuracy

print("Test loss:", round(test_loss, 4))
print("Test accuracy:", round(test_accuracy * 100, 2), "%")
print("Error rate:", round(error_rate * 100, 2), "%")
print("Training latency:", round(training_time, 2), "seconds")
```

```python

import re
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_next_words(sentence):
    sentence = sentence.lower()
    sentence = re.sub(r"[^a-zA-Z\s']", " ", sentence)
    sentence = re.sub(r"\s+", " ", sentence).strip()

    token_list = tokenizer.texts_to_sequences([sentence])[0]

    if len(token_list) == 0:
        print("No known words found. Try another sentence.")
        return

    token_list = token_list[-(max_sequence_len - 1):]

    padded = pad_sequences(
        [token_list],
        maxlen=max_sequence_len - 1,
        padding="pre",
        truncating="pre"
    )

    predictions = model.predict(padded, verbose=0)[0]
    predictions[0] = 0

    top_ids = np.argsort(predictions)[-5:][::-1]

    print("Input sentence:", sentence)
    print("Top 5 next-word predictions:")

    for word_id in top_ids:
        word = tokenizer.index_word.get(int(word_id), "<unknown>")
        probability = predictions[word_id] * 100
        print(f"{word}: {probability:.2f}%")

print("Prediction function defined successfully!")
```
